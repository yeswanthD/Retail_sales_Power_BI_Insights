```python
# Outlier detection using IQR

numeric_columns = [
    'transaction_qty',
    'unit_price',
    'revenue'
]

for col in numeric_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print(f"\nColumn: {col}")
    print("Lower Bound:", round(lower_bound, 2))
    print("Upper Bound:", round(upper_bound, 2))
    print("Outlier Count:", len(outliers))
    print("Outlier Percentage:",
          round(len(outliers) / len(df) * 100, 2), "%")
```

```python
# ==========================================
# 5. EXECUTIVE BUSINESS SUMMARY
# ==========================================

best_store = store_analysis['total_revenue'].idxmax()

best_category = category_analysis['total_revenue'].idxmax()

best_product = product_analysis['total_revenue'].idxmax()

best_month = monthly_analysis.loc[
    monthly_analysis['total_revenue'].idxmax(),
    'month_name'
]

best_hour = hourly_analysis['total_revenue'].idxmax()

lowest_category = category_analysis['total_revenue'].idxmin()

print("========== MAVEN ROASTERS BUSINESS REPORT ==========")

print(f"Total Revenue: ${total_revenue:,.2f}")

print(f"Total Quantity Sold: {total_quantity:,}")

print(f"Total Transactions: {total_transactions:,}")

print(f"Highest Revenue Store: {best_store}")

print(f"Highest Revenue Category: {best_category}")

print(f"Highest Revenue Product: {best_product}")

print(f"Best Performing Month: {best_month}")

print(f"Peak Sales Hour: {best_hour}:00")

print(f"Lowest Revenue Category: {lowest_category}")

print("====================================================")
```

```python
# ======================================
# FINAL EDA RECONCILIATION
# ======================================

import numpy as np

checks = {
    "Store revenue": store_analysis['total_revenue'].sum(),
    "Category revenue": category_analysis['total_revenue'].sum(),
    "Product revenue": product_analysis['total_revenue'].sum(),
    "Monthly revenue": monthly_analysis['total_revenue'].sum(),
    "Hourly revenue": hourly_analysis['total_revenue'].sum()
}

for name, value in checks.items():
    matches = np.isclose(value, total_revenue, atol=0.01)

    print(
        f"{name}: ${value:,.2f} | "
        f"{'PASS' if matches else 'CHECK'}"
    )
```
